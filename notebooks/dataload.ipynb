```python
import os
import subprocess
import time
from pathlib import Path
from kaggle.api.kaggle_api_extended import KaggleApi

# Set the environment variables to point to your correct kaggle.json location
os.environ['KAGGLE_CONFIG_DIR'] = 'C:/Users/Rabbani/.kaggle'
os.environ['KAGGLE_USERNAME'] = 'username'  # Replace with your actual Kaggle username
os.environ['KAGGLE_KEY'] = 'key'        # Replace with your actual Kaggle API key

# Initialize and authenticate
api = KaggleApi()

try:
    api.authenticate()
    print("✅ Authentication successful!")
    
    dataset_identifier = "raddar/chest-xrays-indiana-university"
    download_dir = "D:\AllProjects\med-report-agent\data"
    
    # Get the absolute path for clarity
    absolute_download_path = os.path.abspath(download_dir)
    os.makedirs(download_dir, exist_ok=True)

    print("=" * 60)
    print("📦 DOWNLOAD INFORMATION")
    print("=" * 60)
    print(f"Dataset: {dataset_identifier}")
    print(f"Download location: {absolute_download_path}")
    print(f"Estimated size: ~12-15 GB (compressed)")
    print(f"Uncompressed size: ~15-20 GB")
    print("=" * 60)
    
    # Time estimation based on internet speed
    print("⏰ Estimated download time:")
    print("   - Fast internet (100 Mbps): 20-30 minutes")
    print("   - Average internet (50 Mbps): 40-60 minutes") 
    print("   - Slow internet (10 Mbps): 3-4 hours")
    print("   - Very slow (5 Mbps): 6-8 hours")
    print("=" * 60)
    
    # Confirm before starting
    response = input("Do you want to start the download? (yes/no): ")
    if response.lower() not in ['yes', 'y']:
        print("Download cancelled.")
        exit()
    
    print("🚀 Starting download...")
    print("📊 Download progress will be shown below:")
    print("-" * 40)
    
    # Record start time
    start_time = time.time()
    
    # Download the dataset with progress tracking
    api.dataset_download_files(
        dataset=dataset_identifier,
        path=download_dir,
        unzip=True,
        quiet=False  # This should show progress, but Kaggle API progress can be limited
    )
    
    # Calculate download time
    end_time = time.time()
    total_time = end_time - start_time
    hours, remainder = divmod(total_time, 3600)
    minutes, seconds = divmod(remainder, 60)
    
    print("=" * 60)
    print("✅ DOWNLOAD COMPLETED SUCCESSFULLY!")
    print("=" * 60)
    print(f"📍 Download location: {absolute_download_path}")
    print(f"⏱️  Total download time: {int(hours)}h {int(minutes)}m {int(seconds)}s")
    
    # Show directory contents
    print("\n📁 Contents of download directory:")
    try:
        downloaded_files = os.listdir(download_dir)
        for file in downloaded_files[:10]:  # Show first 10 files
            file_path = os.path.join(download_dir, file)
            file_size = os.path.getsize(file_path) / (1024*1024)  # Size in MB
            print(f"   - {file} ({file_size:.2f} MB)")
        
        if len(downloaded_files) > 10:
            print(f"   ... and {len(downloaded_files) - 10} more files")
            
    except Exception as e:
        print(f"   Could not list directory contents: {e}")
    
    print("=" * 60)
    print("🎉 Dataset is ready for use!")
    print("=" * 60)

except Exception as e:
    print(f"❌ An error occurred: {e}")
    print("\n🔧 Troubleshooting tips:")
    print("1. Make sure you've replaced 'username' and 'key' with your actual Kaggle credentials")
    print("2. Visit the dataset page and accept the terms: https://www.kaggle.com/datasets/raddar/chest-xrays-indiana-university")
    print("3. Check your internet connection")
    print("4. Ensure you have at least 25GB free disk space")
```

```python
import requests

# 1. The Correct API Endpoint for SEARCHING
url = "https://openi.nlm.nih.gov/api/search"

# 2. The HEADERS must specify that we are sending and accepting JSON
headers = {
    "accept": "application/json",
    "Content-Type": "application/json"
}

# 3. The BODY of the request (the most important part!)
# This JSON object defines your search query.
payload = {
    "query": "chest xray pneumonia",  # Your search terms here
    "m": 1,  # Starting index for results (e.g., result #1)
    "n": 10   # Number of results to return (e.g., 10 results)
    # You can add many other parameters here, like "imgTypes": "CT", "age": "40-60"
}

# 4. Make the POST request (NOT a GET request)
try:
    response = requests.post(url, json=payload, headers=headers)
    response.raise_for_status()  # This will raise an exception for 4xx/5xx errors

    # 5. Parse the successful response
    data = response.json()
    print("Request successful!")
    print(f"Total results found: {data.get('total', 0)}")

    # 6. Loop through and print the results
    for result in data.get('list', []):
        print(f"\n--- Result: {result.get('uid')} ---")
        print(f"Title: {result.get('title', 'N/A')}")
        print(f"Caption: {result.get('caption', 'N/A')}")
        
        # Images are often in an 'image' list
        if 'image' in result:
            for img in result['image']:
                # Construct the full URL for the image thumbnail
                img_url = "https://openi.nlm.nih.gov" + img.get('url')
                print(f"Image: {img_url}")

except requests.exceptions.RequestException as e:
    print(f"Request failed: {e}")
except ValueError as e:
    print(f"Failed to decode JSON response: {e}")
    print("Raw response:", response.text)
```
